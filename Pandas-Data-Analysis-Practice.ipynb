```python
import pandas as pd
n=int(input("enter a  no of students"))
rollno=[]
name=[]
age=[]
marks=[]
for i in range(n):
    roll=int(input("enter a rollno"))
    rollno.append(roll)
    names=input("enter a names")
    name.append(names)
    ages=int(input("enter a age"))
    age.append(ages)
    mark=int(input("enterr a marks") )       
    marks.append(mark) 
    
df = pd.DataFrame({
    "Roll No": rollno,
    "Name": name,
    "Age": age,
    "Marks": marks
})
df
```

```python
n=int(input("enter a  no of employee"))
employee=[]
department=[]
salary=[]
for i in range(n):
    name=input("enter a name of employe")
    employee.append(name)
    departments=input("enter a department")
    department.append(departments)
    amount=int(input("enetr a salary"))
    salary.append(amount)
    
df=pd.DataFrame({
    "employee name": employee,
    "department":department,
    "salary":salary
    
})
df
```

```python
df.groupby("department")["salary"].mean()
```

```python
df.groupby("department")["salary"].max()
```

```python
df.groupby("department")["salary"].min()
```
